# Lambda functions

A lambda is a short function without a name: `lambda inputs: output`. It can be stored in a variable, called directly, or passed to another function such as `pandas.Series.apply`, `filter` or `sorted`.

```python
add_ten = lambda x: x + 10
add_ten(5)
```

```python
# the same as a normal function
def add_ten_def(x):
    return x + 10

add_ten_def(5)
```

```python
# called directly, with two inputs
(lambda x, y: x * y)(2, 7)
```

## With pandas

```python
import pandas as pd

df = pd.DataFrame({'a': range(1, 9), 'b': range(4, 12)})
df.apply(lambda col: col + 1)
```

```python
df['label'] = df['a'].apply(lambda n: 'even' if n % 2 == 0 else 'odd')
df
```

## With `filter` and `sorted`

```python
numbers = [1, 2, 3, 4, 5, 6, 8]
list(filter(lambda x: x % 2 == 0, numbers))
```

```python
movies = [('Fay Grim', 2006), ('Alien', 1979), ('Arrival', 2016)]
sorted(movies, key=lambda m: m[1])
```
